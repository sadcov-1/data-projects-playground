```python
df = pd.read_csv(r'C:\Users\ПК\Downloads\train\train.csv')
df.head(5)
```

```python
df.isnull().sum()
```

```python
df["Ticket"].dtype
```

```python
bygen = df.groupby('Sex')['Survived'].mean()
bygen 
```

```python
bypclass = df.groupby('Pclass')['Survived'].mean()
bypclass
```

```python
df['AgeGroup'] = pd.cut(df['Age'], bins = [0, 40, 55, float('inf')], labels =['Young', 'Medium', 'Old'])
df
```

```python
byagegroup = df.groupby('AgeGroup')['Survived'].mean()
byagegroup
```

```python
plt.figure(figsize=(5, 7))
byagegroup.plot(kind = 'bar')
plt.show()
```

```python
plt.figure(figsize = (5,7))
bypclass.plot(kind = 'bar')
plt.show()
```

```python
plt.figure(figsize=(4,6))
bygen.plot(kind='bar')
plt.show()
```
