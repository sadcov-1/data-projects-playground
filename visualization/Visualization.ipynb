```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
tips.groupby('sex')['tip'].mean()
```

```python
tips.groupby('sex')['total_bill'].mean()
```

```python
df = tips.groupby('day')['size'].sum()
```

```python
plt.figure(figsize=(5,7))
df.plot(kind='bar')
plt.show()
```

From this barchart we can see that most people come on saturday and sunday.

```python
plt.figure(figsize=(2,5))
plt.hist(tips['sex'])
plt.show()
```

Most clients are male.

```python
plt.figure(figsize=(10,3))
plt.boxplot(tips['tip'], vert=False)
plt.title('Tips')
plt.grid()
plt.xticks(range(0, 11, 1))
plt.show()
```

Median of tips is about 2.8$, 25% of tips is less or equal to 1$, 75% of tips is less or equal to around 3.6$, highest tip was 10$. 

```python
plt.scatter(tips['total_bill'], tips['tip'])

plt.xlabel('Total bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')

plt.show()
```

```python
corr = tips[['total_bill', 'tip', 'size']].corr()
```

```python
sns.heatmap(corr, annot=True)
plt.show()
```

From the heatmap and scatterplot, we can see the correlations between tips, group size, and total bill. Tips have the strongest correlation with total bill (0.68), while total bill is also moderately correlated with group size (0.60). This is logical because larger groups tend to have higher total bills.
