```python
import pandas as pd 
import matplotlib.pyplot as plt 
```

What are the total sales quantities for each drug category (ATC code)?

```python
df1 = df.drop(columns='datum').sum()
```

```python
df1.sort_values(ascending=False)
```

Which individual drug brands have the highest total sales?

```python
df1.sort_values(ascending=False).head(1)
```

Which three drugs have the highest sales in January 2015, July 2016, September 2017.

```python
df['datum'] = pd.to_datetime(df['datum'])
```

```python
jan = df[(df['datum'].dt.year == 2015) & (df['datum'].dt.month == 1)]
```

```python
jul = df[(df['datum'].dt.year == 2016) & (df['datum'].dt.month == 7)]
```

```python
sep = df[(df['datum'].dt.year == 2017) & (df['datum'].dt.month == 9)]
```

```python
jan.drop(columns ='datum').sum().sort_values(ascending = False).head(3)
```

```python
jul.drop(columns ='datum').sum().sort_values(ascending = False).head(3)
```

```python
sep.drop(columns ='datum').sum().sort_values(ascending = False).head(3)
```

Which drug has sold the most often in 2017?


```python
mostoft = df[df['datum'].dt.year == 2017]
mostoft.drop(columns='datum').sum().sort_values(ascending=False).head(1)
```
