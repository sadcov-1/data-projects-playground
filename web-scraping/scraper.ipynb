```python
import requests
from bs4 import BeautifulSoup
import csv
```

```python
url = 'https://realpython.github.io/fake-jobs/'
response = requests.get(url)
```

```python
response.status_code
```

```python
print(response.text[:1000])
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
type(soup)
```

```python
jobs = soup.select('div.card-content')
len(jobs)
```

```python
jobs[0]
```

```python
title = jobs[0].select_one('h2.title.is-5')
title.get_text(strip=True)
```

```python
company = jobs[0].select_one('h3.company')
company.get_text(strip=True)
```

```python
location = jobs[0].select_one('p.location')
location
```

```python
location.get_text(strip=True)
```

```python
link = jobs[0].select('a.card-footer-item')[1]
link
```

```python
link.get('href')
```

```python
jobs_data = []

for job in jobs:
    title = job.select_one("h2.title.is-5").get_text(strip=True)
    company = job.select_one("h3.company").get_text(strip=True)
    location = job.select_one("p.location").get_text(strip=True)
    link = job.select("a.card-footer-item")[1].get("href")

    jobs_data.append({
        "title": title,
        "company": company,
        "location": location,
        "url": link
    })
```

```python
jobs_data
```

```python
with open("go_to.csv", "w", newline="", encoding="utf-8") as file:
    fieldnames = ["title", "company", "location", "url"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(jobs_data)
```
